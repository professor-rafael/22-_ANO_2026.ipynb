JOGO- ADIVINHE O NÚMERO SECRETO

```python
import random

numero_secreto = random.randint(1, 10)

tentativa_usuario = int(input("Adivinhe o número entre 1 e 10:"))

if tentativa_usuario == numero_secreto:
    print("Parabéns! Você acertou! 🎉")
else:
    print("Não foi dessa vez! O número secreto era:", numero_secreto, "😂")
```
